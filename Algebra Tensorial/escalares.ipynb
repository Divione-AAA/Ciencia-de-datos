```python
a = tf.constant(10)
b = tf.constant(2)
c = tf.constant(9)
c = tf.cast(c,tf.float32)
```

```python
#operaciones con tensores
print(a + b)
print(a - b)
print(a ** b)
print(a - b)
print(a * b)
print(a / b)
print(tf.sqrt(tf.constant(c)))
```

```python
#Ejecricio 1
a = tf.constant(5)
b = tf.constant(-8)
c = tf.constant(3.14)
d = tf.constant(True)
e = tf.constant("Tensorflow")
print(a)
print(a.dtype)
print(a.shape)
print(b)
print(b.dtype)
print(b.shape)
print(c)
print(c.dtype)
print(c.shape)
print(d)
print(d.dtype)
print(d.shape)
print(e)
print(e.dtype)
print(e.shape)
```

```python
#Ejercicio 2
a = tf.constant(15)
b = tf.constant(4)


print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**b)
print(tf.math.mod(a,b))
print(tf.maximum(a,b))
print(tf.minimum(a,b))
```
