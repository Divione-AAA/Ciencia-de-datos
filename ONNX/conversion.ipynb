```python
from tensorflow.keras.models import load_model
from onnxruntime.quantization import quantize_dynamic, QuantType
import tensorflow as tf
import tf2onnx
import onnx
```

```python
model = load_model('best_malaria_cnn.h5')
```

```python
model_proto = tf2onnx.convert.from_keras(
    model,
    opset=13
)
```

```python
spec = (tf.TensorSpec((None, 224, 224, 3), tf.float32, name="input"),)
onnx_model, _ = tf2onnx.convert.from_keras(model,input_signature=spec, opset=13)
onnx.save(onnx_model, "modelo_convertido.onnx")
```

```python
model_fp32 = '/model.onxx'
model_quant = '/modelq.onxx'
quantize_model = quantize_dynamic(model_fp32, model_quant, weight_type = QuantType.QInt16)
```
