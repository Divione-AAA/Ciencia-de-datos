```python
import tensorflow as tf
from tensorflow.keras import layers
from Transformersblock import TransformerBlock 
```

```python
inputs = layers.Input(shape=(224,224,3))

# Patches: divide la imagen en patches 16 x 16
patches = tf.image.extract_patches(
    images=inputs,
    sizes=[1,16,16,1],
    strides=[1,16,16,1],
    rates=[1,1,1,1],
    padding="VALID"
)

# Flatten + Embedding: cada patch se proyecta sobre una dimension
patches = layers.Reshape((-1, patches.shape[-1]))(patches)
x = layers.Dense(128)(patches)

# Transformer: se aplican multiples bloques de atencion
x = TransformerBlock(128, 4, 256)(x)

# Pooling
x = layers.GlobalAveragePooling1D()(x)

# Output
outputs = layers.Dense(3, activation="softmax")(x)
model = tf.keras.Model(inputs, outputs)
```
